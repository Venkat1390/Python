```python
def Number():
    num = int(input("Enter a Number:"))

    if num % 2 == 0:
        print(num, "is an Even Number")
    else:
        print(num, "is an Odd Number")

Number()
```

```python
def Eligible():
    gender = input("Enter your Gender:")
    age = int(input("Enter your Age:"))

    if gender == "Male":
        if age >= 21:
            print("ELIGIBLE")
        else:
            print("NOT ELIGIBLE")

    elif gender == "Female":
        if age >= 18:
            print("ELIGIBLE")
        else:
            print("NOT ELIGIBLE")

Eligible()
```
